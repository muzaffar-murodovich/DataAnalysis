```python
import numpy as np
import scipy as sp
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
import os
import urllib.request

url = 'https://raw.githubusercontent.com/Dilmurod09/Tashkent-real-estate-ml/main/data/uybor.xlsx'
os.makedirs('input', exist_ok=True)
urllib.request.urlretrieve(url, 'input/uybor.xlsx')

df_xlsx = pd.read_excel('input/uybor.xlsx')
df_xlsx.to_csv('input/uybor.csv', index=False)
df_xlsx.shape
```
